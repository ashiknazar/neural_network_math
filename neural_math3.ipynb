# Forward Propagation

## Inputs
a0 = 170  
a1 = 80  

## Weights
w1 = 0.4   (a0 → h1)  
w2 = 0.7   (a1 → h1)  

w3 = 0.5   (a0 → h2)  
w4 = 0.2   (a1 → h2)  

w5 = 0.5   (h1 → o1)  
w6 = 0.5   (h2 → o1)  

## Biases
b1 = 0.11 (h1)  
b2 = 0.43 (h2)  
b3 = 0.11 (output)

---

## Hidden Layer

### h1
z1 = (a0 · w1) + (a1 · w2) + b1  
   = (170 × 0.4) + (80 × 0.7) + 0.11  
   = 124.11  

a2 = ReLU(z1) = 124.11  

---

### h2
z2 = (a0 · w3) + (a1 · w4) + b2  
   = (170 × 0.5) + (80 × 0.2) + 0.43  
   = 101.43  

a3 = ReLU(z2) = 101.43  

---

## Output Layer

z3 = (a2 · w5) + (a3 · w6) + b3  
   = (124.11 × 0.5) + (101.43 × 0.5) + 0.11  
   = 112.88  

a4 = ReLU(z3) = 112.88  

---

# Backward Propagation

## Target
let y = 113  

Loss:
L = (1/2)(a4 - y)²  

---

## Output Layer

dL/da4 = (a4 - y) = (112.88 - 113) = -0.12  

ReLU'(z3) = 1  

dL/dz3 = -0.12  

---

## Output Weights

dL/dw5 = dL/dz3 · a2 = -0.12 × 124.11 = -14.89  
dL/dw6 = dL/dz3 · a3 = -0.12 × 101.43 = -12.17  

dL/db3 = -0.12  

---

## Hidden Layer

### h1

dL/da2 = dL/dz3 · w5 = -0.12 × 0.5 = -0.06  

ReLU'(z1) = 1  

dL/dz1 = -0.06  

---

### h2

dL/da3 = dL/dz3 · w6 = -0.12 × 0.5 = -0.06  

ReLU'(z2) = 1  

dL/dz2 = -0.06  

---

## Input Weights

### h1

dL/dw1 = dL/dz1 · a0 = -0.06 × 170 = -10.2  
dL/dw2 = dL/dz1 · a1 = -0.06 × 80 = -4.8  

---

### h2

dL/dw3 = dL/dz2 · a0 = -0.06 × 170 = -10.2  
dL/dw4 = dL/dz2 · a1 = -0.06 × 80 = -4.8  

---

## Bias Gradients

dL/db1 = -0.06  
dL/db2 = -0.06
