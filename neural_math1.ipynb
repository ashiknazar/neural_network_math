# Neural Network (From Given Diagram)

## 1. Structure

- Inputs: $a_0$ (height), $a_1$ (weight)
- Hidden layer (ReLU): $Z_0, Z_1, Z_2$
- Outputs: $Z_3, Z_4$ (no activation)

---

## 2. Weights (from image)

### Input → Hidden
$$
\begin{aligned}
w_1 &= 0.2 \quad (a_0 \rightarrow Z_0) \\
w_2 &= 0.5 \quad (a_0 \rightarrow Z_1) \\
w_3 &= 0.7 \quad (a_0 \rightarrow Z_2) \\
w_4 &= 0.5 \quad (a_1 \rightarrow Z_0) \\
w_5 &= 0.8 \quad (a_1 \rightarrow Z_1) \\
w_6 &= 0.4 \quad (a_1 \rightarrow Z_2)
\end{aligned}
$$

---

### Hidden → Output
$$
\begin{aligned}
w_7 &= 0.6 \quad (Z_0 \rightarrow Z_3) \\
w_8 &= 0.7 \quad (Z_0 \rightarrow Z_4) \\
w_9 &= 0.11 \quad (Z_1 \rightarrow Z_3) \\
w_{10} &= 0.5 \quad (Z_1 \rightarrow Z_4) \\
w_{11} &= 0.4 \quad (Z_2 \rightarrow Z_3) \\
w_{12} &= 0.5 \quad (Z_2 \rightarrow Z_4)
\end{aligned}
$$

---

## 3. Biases
$$
b_0, b_1, b_2 \text{ (hidden)}, \quad b_3, b_4 \text{ (output)}
$$

Assume:
$$
b_0 = b_1 = b_2 = b_3 = b_4 = 0.1
$$

---

# 4. Input (Sample 1)

$$
a_0 = 170,\quad a_1 = 80
$$

---

# 5. Forward Propagation

## Hidden Layer (Pre-activation)

$$
Z_0 = a_0 w_1 + a_1 w_4 + b_0
= (170)(0.2) + (80)(0.5) + 0.1 = 74.1
$$

$$
Z_1 = a_0 w_2 + a_1 w_5 + b_1
= (170)(0.5) + (80)(0.8) + 0.1 = 149.1
$$

$$
Z_2 = a_0 w_3 + a_1 w_6 + b_2
= (170)(0.7) + (80)(0.4) + 0.1 = 151.1
$$

---

## Activation (ReLU)

$$
a_2 = \text{ReLU}(Z_0) = 74.1
$$

$$
a_3 = 149.1,\quad a_4 = 151.1
$$

---

## Output Layer

$$
Z_3 = a_2 w_7 + a_3 w_9 + a_4 w_{11} + b_3
$$

$$
= (74.1)(0.6) + (149.1)(0.11) + (151.1)(0.4) + 0.1
$$

$$
= 44.46 + 16.401 + 60.44 + 0.1 = 121.401
$$

---

$$
Z_4 = a_2 w_8 + a_3 w_{10} + a_4 w_{12} + b_4
$$

$$
= (74.1)(0.7) + (149.1)(0.5) + (151.1)(0.5) + 0.1
$$

$$
= 51.87 + 74.55 + 75.55 + 0.1 = 202.07
$$

---

## Outputs

$$
a_5 = Z_3 = 121.401,\quad a_6 = Z_4 = 202.07
$$

---

# 6. Loss

Assume:
$$
y_1 = 130,\quad y_2 = 210
$$

$$
L = \frac{1}{2}(a_5 - y_1)^2 + \frac{1}{2}(a_6 - y_2)^2
$$

---

# 7. Backpropagation

## Output Gradients

$$
\frac{\partial L}{\partial Z_3} = a_5 - y_1 = -8.599
$$

$$
\frac{\partial L}{\partial Z_4} = -7.93
$$

---

## Gradients for Hidden Layer

### Chain Rule

$$
\frac{\partial L}{\partial a_2} =
\frac{\partial L}{\partial Z_3} w_7 +
\frac{\partial L}{\partial Z_4} w_8
$$

$$
= (-8.599)(0.6) + (-7.93)(0.7) = -10.71
$$

---

$$
\frac{\partial L}{\partial a_3} =
(-8.599)(0.11) + (-7.93)(0.5)
$$

$$
= -0.946 + (-3.965) = -4.911
$$

---

$$
\frac{\partial L}{\partial a_4} =
(-8.599)(0.4) + (-7.93)(0.5)
$$

$$
= -3.4396 + (-3.965) = -7.4046
$$

---

## ReLU Backprop

$$
\frac{\partial a}{\partial Z} =
\begin{cases}
1 & Z > 0 \\
0 & Z \le 0
\end{cases}
$$

(All positive → gradient = 1)

---

## Input Layer Gradients

$$
\frac{\partial L}{\partial w_1} =
\frac{\partial L}{\partial Z_0} \cdot a_0
= (-10.71)(170) = -1820.7
$$

$$
\frac{\partial L}{\partial w_4} =
(-10.71)(80) = -856.8
$$

---

# 8. Parameter Update

Learning rate:
$$
\eta = 0.001
$$

$$
w_1^{new} = 0.2 - (0.001)(-1820.7) = 2.0207
$$

$$
w_4^{new} = 0.5 - (0.001)(-856.8) = 1.3568
$$

---

# 9. Chain Rule Dependency Summary

$$
L \rightarrow Z_3, Z_4
$$

$$
Z_3, Z_4 \rightarrow a_2, a_3, a_4
$$

$$
a_2, a_3, a_4 \rightarrow Z_0, Z_1, Z_2
$$

$$
Z_0, Z_1, Z_2 \rightarrow w_1, w_2, w_3, w_4, w_5, w_6
$$

---

# 10. Key Insight

- Forward: linear → ReLU → linear  
- Backward: gradients flow **right → left**
- ReLU simplifies derivatives
- Large inputs → unstable gradients → normalize in practice

$$
170 \rightarrow 1.70,\quad 80 \rightarrow 0.80
$$

$$
\text{ReLU}(x) = \max(0, x)

$$
\begin{aligned}
w_1 &= 0.2,\quad w_2 = 0.5,\quad w_3 = 0.7 \\

$$
\begin{aligned}

# Backpropagation (Line-by-Line Derivation)

## Given (from forward pass)

$$
a_5 = 121.401,\quad a_6 = 202.07
$$

$$
y_1 = 130,\quad y_2 = 210
$$

---

# STEP 1: Loss Gradient w.r.t Outputs

$$
\frac{\partial L}{\partial a_5}
= \frac{\partial}{\partial a_5} \left( \frac{1}{2}(a_5 - y_1)^2 \right)
= (a_5 - y_1)
$$

$$
= 121.401 - 130 = -8.599
$$

---

$$
\frac{\partial L}{\partial a_6}
= (a_6 - y_2)
= 202.07 - 210 = -7.93
$$

---

# STEP 2: Output Layer (ReLU Backprop)

$$
a_5 = \text{ReLU}(Z_3)
$$

$$
\frac{\partial a_5}{\partial Z_3} =
\begin{cases}
1 & Z_3 > 0 \\
0 & Z_3 \le 0
\end{cases}
= 1
$$

---

$$
\frac{\partial L}{\partial Z_3}
=
\frac{\partial L}{\partial a_5}
\cdot
\frac{\partial a_5}{\partial Z_3}
$$

$$
= (-8.599)(1) = -8.599
$$

---

Similarly:

$$
\frac{\partial L}{\partial Z_4}
= (-7.93)(1) = -7.93
$$

---

# STEP 3: Gradients w.r.t Output Weights

### For $w_7$

$$
Z_3 = a_2 w_7 + a_3 w_9 + a_4 w_{11} + b_3
$$

$$
\frac{\partial Z_3}{\partial w_7} = a_2
$$

$$
\frac{\partial L}{\partial w_7}
=
\frac{\partial L}{\partial Z_3}
\cdot
\frac{\partial Z_3}{\partial w_7}
$$

$$
= (-8.599)(74.1) = -637.18
$$

---

### For $w_8$

$$
\frac{\partial Z_4}{\partial w_8} = a_2
$$

$$
\frac{\partial L}{\partial w_8}
= (-7.93)(74.1) = -587.61
$$

---

### For $w_9$

$$

$$

$$
\frac{\partial L}{\partial w_9}
= (-8.599)(149.1) = -1281.69
$$

---
